# Arechi Urban Study - Analisi Geospaziale

Questo notebook raccoglie le analisi spaziali quantitative per lo studio urbanistico e la proposta di rigenerazione del **Piazzale Gipo Viani** e dello **Stadio Arechi** a Salerno.

## Indice delle Analisi:
1. **Caricamento dei dati spaziali (OSM)**
2. **Analisi delle destinazioni d'uso (Land Use)**
3. **Analisi di accessibilità e sosta (Isocrone)**
4. **Analisi dei flussi pedonali e congestione**

## 1. Caricamento dei dati e Viabilità
Carichiamo la rete stradale e verifichiamo la consistenza dei dati scaricati tramite OSMnx.

```python
import geopandas as gpd
import osmnx as ox
import matplotlib.pyplot as plt
import pandas as pd
import os

# Definiamo i percorsi dei dati
data_dir = "../data/raw/osm"
streets_path = os.path.join(data_dir, "streets_drive.gpkg")

if os.path.exists(streets_path):
    streets = gpd.read_file(streets_path, layer="edges")
    print(f"Rete carrabile caricata: {len(streets)} segmenti stradali.")
    
    # Visualizzazione della rete
    fig, ax = plt.subplots(figsize=(8, 8))
    streets.plot(ax=ax, color='gray', linewidth=0.8)
    plt.title("Rete Carrabile dell'Area di Studio")
    plt.axis('off')
    plt.show()
else:
    print("File delle strade non trovato. Esegui prima il download.")
```

## 2. Analisi delle destinazioni d'uso (Land Use)
Analizziamo la distribuzione dell'edificato per capire quali funzioni urbanistiche prevalgono nell'area di studio.

```python
buildings_path = os.path.join(data_dir, "buildings.gpkg")

if os.path.exists(buildings_path):
    buildings = gpd.read_file(buildings_path)
    
    # Raggruppamento e pulizia
    building_types = buildings['building'].value_counts()
    print("--- Edifici per tipologia OSM ---")
    print(building_types.head(10))
    
    # Visualizzazione degli edifici
    fig, ax = plt.subplots(figsize=(8, 8))
    buildings.plot(ax=ax, color='#2c3e50', alpha=0.9)
    plt.title("Footprints degli Edifici")
    plt.axis('off')
    plt.show()
else:
    print("File degli edifici non trovato.")
```

## 3. Analisi dell'Accessibilità e della Sosta
Visualizziamo i buffer di accessibilità pedonale (200m, 500m, 800m) dal centro dello Stadio Arechi e la localizzazione dei parcheggi.

```python
processed_dir = "../data/processed"
parking_path = os.path.join(processed_dir, "parking_processed.geojson")
buffers_path = os.path.join(processed_dir, "stadium_buffers.geojson")

if os.path.exists(parking_path) and os.path.exists(buffers_path):
    parkings = gpd.read_file(parking_path)
    buffers = gpd.read_file(buffers_path)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    # Disegno dei buffer
    buffers.plot(ax=ax, cmap='Set2', alpha=0.3, edgecolor='black', linewidth=1)
    # Disegno dei parcheggi
    parkings.plot(ax=ax, color='#2c3e50', markersize=parkings['estimated_capacity']*2, alpha=0.7)
    
    plt.title("Fasce di Accessibilità e Capacità Parcheggi")
    plt.axis('off')
    plt.show()
    
    print(f"Capacità totale di sosta rilevata: {parkings['estimated_capacity'].sum()} posti auto.")
else:
    print("File elaborati per l'accessibilità non trovati. Esegui prima scripts/process_gis.py.")
```

## 4. Conclusioni dello Studio dello Stato di Fatto
- L'area intorno allo **Stadio Arechi** è caratterizzata da un'elevatissima disponibilità di parcheggi a raso (asfalto nudo), che favorisce l'effetto **isola di calore urbana**.
